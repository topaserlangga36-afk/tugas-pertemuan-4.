```python
import requests
import pandas as pd
import json
import hashlib
import time
from pathlib import Path
from datetime import datetime
from zoneinfo import ZoneInfo

NAMA = "I Kadek Topas Erlangga"
NIM = "250101010328"

BASE_DIR = Path.cwd()

if BASE_DIR.name == "notebooks":
    BASE_DIR = BASE_DIR.parent

RAW_DIR = BASE_DIR / "data" / "raw"
DOCS_DIR = BASE_DIR / "docs"

RAW_DIR.mkdir(parents=True, exist_ok=True)
DOCS_DIR.mkdir(parents=True, exist_ok=True)

API_URL = "https://api.open-meteo.com/v1/forecast"
TIMEZONE = "Asia/Makassar"
```

```python
destinasi = [
    {
        "lokasi": "Pantai Kuta",
        "latitude": -8.7184,
        "longitude": 115.1686
    },
    {
        "lokasi": "Tanah Lot",
        "latitude": -8.6212,
        "longitude": 115.0868
    },
    {
        "lokasi": "Ulun Danu Beratan",
        "latitude": -8.2753,
        "longitude": 115.1669
    },
    {
        "lokasi": "Ubud Palace",
        "latitude": -8.5069,
        "longitude": 115.2625
    },
    {
        "lokasi": "Garuda Wisnu Kencana",
        "latitude": -8.8106,
        "longitude": 115.1669
    },
    {
        "lokasi": "Kintamani",
        "latitude": -8.2420,
        "longitude": 115.3750
    }
]

pd.DataFrame(destinasi)
```

```python
data_api = []

# Lakukan request ke API untuk setiap destinasi
for tempat in destinasi:
    params = {
        "latitude": tempat["latitude"],
        "longitude": tempat["longitude"],
        "hourly": "temperature_2m,precipitation,cloud_cover,wind_speed_10m",
        "timezone": TIMEZONE
    }
    response = requests.get(API_URL, params=params)
    if response.status_code == 200:
        data_api.append(response.json())
    else:
        print(f"Gagal mengambil data untuk {tempat['lokasi']}")

records = []

for tempat, hasil in zip(destinasi, data_api):

    hourly = hasil["hourly"]

    waktu = hourly["time"]
    suhu = hourly["temperature_2m"]
    hujan = hourly["precipitation"]
    awan = hourly["cloud_cover"]
    angin = hourly["wind_speed_10m"]

    for i in range(len(waktu)):

        records.append({
            "lokasi": tempat["lokasi"],
            "latitude": tempat["latitude"],
            "longitude": tempat["longitude"],
            "waktu_wita": waktu[i],
            "suhu_c": suhu[i],
            "curah_hujan_mm": hujan[i],
            "tutupan_awan_persen": awan[i],
            "kecepatan_angin_kmh": angin[i]
        })

df = pd.DataFrame(records)

print("Jumlah record:", len(df))
print("Jumlah kolom:", len(df.columns))

display(df.head(10))
```

```python
waktu_akuisisi = datetime.now(
    ZoneInfo(TIMEZONE)
)

tanggal = waktu_akuisisi.strftime("%Y%m%d")

nama_file = f"openmeteo_bali_{tanggal}.csv"

csv_path = RAW_DIR / nama_file

if csv_path.exists():
    nama_file = (
        f"openmeteo_bali_{tanggal}_"
        f"{waktu_akuisisi.strftime('%H%M%S')}.csv"
    )
    csv_path = RAW_DIR / nama_file

df.to_csv(
    csv_path,
    index=False,
    encoding="utf-8"
)

print("Dataset berhasil disimpan!")
print("Lokasi:", csv_path)
```

```python
ukuran_file = csv_path.stat().st_size

hash_md5 = hashlib.md5(
    csv_path.read_bytes()
).hexdigest()

provenance = {
    "nama_berkas": str(csv_path.relative_to(BASE_DIR)),
    "ukuran_byte": ukuran_file,
    "hash_md5": hash_md5,
    "sumber": "Open-Meteo Forecast API",
    "url": API_URL,
    "lisensi": "CC BY 4.0",
    "atribusi": "Weather data by Open-Meteo.com",
    "waktu_akuisisi": waktu_akuisisi.isoformat(),
    "metode": "API",
    "jumlah_record": len(df),
    "jumlah_kolom": len(df.columns),
    "nama_pengambil": NAMA,
    "nim_pengambil": NIM
}

with open(
    DOCS_DIR / "provenance.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        provenance,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Provenance berhasil dibuat!")
```
